```python
# Print Odd Numbers from 1 to 10
def print_odd_number():
    for i in range(1,11):
        if i%2!=0:
            print(i)
print_odd_number()
```

```python
# Print Even Numbers from 1 to 10
def print_even_number():
    for i in range(1,11):
        if i % 2 == 0:
            print(i)
print_even_number()
```
